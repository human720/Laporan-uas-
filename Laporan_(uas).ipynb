```python
# =========================
# LAPORAN PENGGUNAAN
# =========================
def laporan_penggunaan():
    print("\n=== LAPORAN PENGGUNAAN ===")
    with koneksi() as conn:
        cur = conn.cursor()

        cur.execute("SELECT COUNT(*) FROM booking")
        total_booking = cur.fetchone()[0]

        cur.execute("""
            SELECT COUNT(*) FROM booking
            WHERE status_booking NOT IN ('DIBATALKAN', 'SELESAI')
        """)
        booking_aktif = cur.fetchone()[0]

        cur.execute("SELECT COUNT(*) FROM booking WHERE status_booking = 'SELESAI'")
        booking_selesai = cur.fetchone()[0]

        cur.execute("""
            SELECT COALESCE(SUM(total_harga), 0)
            FROM booking WHERE status_bayar = 'LUNAS'
        """)
        pendapatan = cur.fetchone()[0]

        cur.execute("""
            SELECT f.nama_fasilitas, COUNT(b.id) AS jumlah
            FROM fasilitas f
            LEFT JOIN booking b ON f.id = b.id_fasilitas
            GROUP BY f.id
            ORDER BY jumlah DESC LIMIT 1
        """)
        populer = cur.fetchone()

    print(f"Total booking       : {total_booking}")
    print(f"Booking aktif       : {booking_aktif}")
    print(f"Booking selesai     : {booking_selesai}")
    print(f"Total pendapatan    : Rp{pendapatan:,}")
    if populer and populer[1] > 0:
        print(f"Fasilitas terpopuler: {populer[0]} ({populer[1]} booking)")
    else:
        print("Fasilitas terpopuler: Belum ada data")
```
